1. Write a program that:
  *   Asks the user for a number.
  *   Uses a for loop to print the multiplication table of that number up to 10.

```python
# Asks the user for a number
num = int(input("Enter a number: "))

# Prints the multiplication table of the user's number up to 10
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

2. Write a program that:
*   Defines separate functions for add, subtract, multiply, and divide.
*   Asks the user to input two numbers and choose an operation.
*   Calls the correct function and prints the result.

```python
# Defines a separate functions for each mathematical operation
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Cannot divide by zero"
    return a / b

# Asks the user for two numbers and an operation
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("Choose an operation: +, -, *, /")
operation = input("Enter operation: ")

# Call the correct function based on the chosen operation
if operation == "+":
    result = add(num1, num2)
elif operation == "-":
    result = subtract(num1, num2)
elif operation == "*":
    result = multiply(num1, num2)
elif operation == "/":
    result = divide(num1, num2)
else:
    result = "Invalid operation"

print(f"\nResult: {result}")
```
