1: Write a program that:
*   Stores the prices of 5 grocery items in a list.
*   Asks the user to enter the quantities of each item.
*   Calculates and prints the total cost of all items.


```python
#Stores the prices of the 5 grocery items
items = ["Apples", "Eggs", "Bread", "Milk", "Cheese"]
prices = [0.50, 4.75, 3.50, 6.20, 2.25]

quantities = []

#Asks the user for the quantity of each item
for i in range(len(items)):
    qty = int(input(f"Enter quantity for {items[i]} (${prices[i]:.2f} each): "))
    quantities.append(qty)

#Calculate total cost
total_cost = 0
for i in range(len(items)):
    total_cost += prices[i] * quantities[i]
print(f"\nTotal cost of all items: ${total_cost:.2f}")
```

2: Write a program that:
* Takes a sentence as input.
* Splits it into words and stores them in a set.
* Prints the set of unique words and the number of unique words.

```python
#Takes the users sentence as an input
sentence = input("Please enter a sentence: ")

#Splits sentence into words and stores as a set
words = sentence.split()
unique_words = set(words)

#prints the unique word set, and the number of unique words
print(f"\nSet of unique words: {unique_words}")
print(f"Number of unique words: {len(unique_words)}")
```

3: Write a program that:
* Creates a dictionary with 3 people’s names as keys and their favorite fruit as values.
* Prints the dictionary.
* Asks the user to type a person’s name and prints that person’s favorite fruit.

```python
# Create dictionary with names as keys and favorite fruit as values
favorite_fruits = {
    "Ashleigh": "Tomato",
    "Eric": "Cherry",
    "Kenny": "Raspberry"
}

# Print the dictionary
print("Favorite fruits dictionary:")
print(favorite_fruits)

# Ask user for a name and print that person's favorite fruit
name = input("\nEnter a person's name: ")

if name in favorite_fruits:
    print(f"{name}'s favorite fruit is {favorite_fruits[name]}.")
else:
    print(f"Sorry, I don't have fruit info for {name}.")
```
